```python
import pandas as pd
from pymongo import MongoClient
from dotenv import load_dotenv
import os

load_dotenv()

# Connect to MongoDB
mongo_client = MongoClient(os.getenv('MONGO_URI'))
db = mongo_client[os.getenv('MONGO_DB_NAME')]
collection = db[os.getenv('MONGO_METADATA_COLLECTION')]
```

```python

# Load the CSV file
df = pd.read_csv('/home/vikasgoyal/Desktop/Projects/intelligent_medicine_strip_analysis_system/PR-MSD/packaged-drug-detection-module/assets/Tablet_Config.csv')

# Rename columns (example)
df.rename(columns={
    'Tablet Name': 'medicine_name',
    'Total Tablets': 'total_tablets',
    'Area in cm2': 'area_of_strip (in cm^2)',
}, inplace=True)

# Convert DataFrame to dictionary with medicine name as key and total_tablets and area_of_strip (in cm^2) as attributes
medicine_dict = df.set_index('medicine_name')[['total_tablets', 'area_of_strip (in cm^2)']].to_dict('index')

# Insert the dictionary into MongoDB
for medicine_name, attributes in medicine_dict.items():
    collection.insert_one({'medicine_name': medicine_name, 'attributes': attributes})

print("Data inserted successfully!")
```

```python

# Update the column name in MongoDB
collection.update_many(
    {},
    {"$rename": {"attributes.area_of_strip (in cm^2)": "attributes.area_of_strip"}}
)

print("Column name updated successfully!")
```

```python
collection.update_many(
    {},
    {"$set": {"category": "strip"}}
)

print("Category field added successfully to all documents!")
```
